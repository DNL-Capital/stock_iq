# Kiểm thử `macro_economic`

Các chỉ số kinh tế vĩ mô Việt Nam theo năm, quý và tháng.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

```python
from stockiq import macro_economic
import pandas as pd

# GDP theo năm
df = macro_economic.get_general("GDP", "years", "2021", "2024")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import macro_economic
import pandas as pd

# GDP theo quý bằng ID
df = macro_economic.get_general(43, "quarters", "2020", "2023")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import macro_economic
import pandas as pd

# CPI theo tháng
df = macro_economic.get_general("CPI", "months", "2023-01", "2023-06")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```
