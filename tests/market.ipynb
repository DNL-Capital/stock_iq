# Kiểm thử `market`

Dữ liệu toàn thị trường, không cần một mã cổ phiếu cố định.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

```python
from stockiq import market
import pandas as pd

# Bảng giá nhiều mã
df = market.get_price_board(symbols=["ACB", "FPT", "E1VFVN30"])
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import market
import pandas as pd
from datetime import date, timedelta

# Lịch sự kiện 30 ngày gần nhất
end = date.today()
start = end - timedelta(days=30)
df = market.get_event_calendar(start=start.isoformat(), end=end.isoformat())
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import market
import pandas as pd
from datetime import date, timedelta

# Lọc lịch sự kiện cổ tức và đại hội cổ đông
end = date.today()
start = end - timedelta(days=30)
df = market.get_event_calendar(start=start.isoformat(), end=end.isoformat(), event_codes=["DIV", "AGME"])
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```
