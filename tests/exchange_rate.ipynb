# Kiểm thử `exchange_rate.get_vcb`

Tỷ giá ngoại tệ Vietcombank theo ngày.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

```python
from stockiq import exchange_rate
import pandas as pd

# Tỷ giá hôm nay
df = exchange_rate.get_vcb()
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```

```python
from stockiq import exchange_rate
import pandas as pd

# Tỷ giá ngày lịch sử
df = exchange_rate.get_vcb("2023-12-26")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```
