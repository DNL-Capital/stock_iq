# Kiểm thử `fund`

Danh sách quỹ mở, NAV và cơ cấu danh mục đầu tư.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

```python
from stockiq import fund
import pandas as pd

# Danh sách quỹ
df = fund.get_listing()
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import fund
import pandas as pd

# Lịch sử NAV của BMFF
df = fund.get_nav_history(fund="BMFF", start="2023-03-01", end="2024-01-01")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```
