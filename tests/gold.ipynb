```python
from stockiq import gold
import pandas as pd

# Giá vàng hôm nay
df = gold.get_sjc()
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```
