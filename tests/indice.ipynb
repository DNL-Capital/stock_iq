# Kiểm thử `indice`

Lịch sử các chỉ số chứng khoán Việt Nam.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

```python
from stockiq import indice
import pandas as pd

# Chỉ số VNINDEX
df = indice.get_market(indice="VNINDEX", start="2024-01-02", end="2024-01-10")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```
