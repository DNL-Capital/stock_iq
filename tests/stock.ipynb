# Kiểm thử `stock`

Dữ liệu cổ phiếu Việt Nam: thị trường, tài chính, cổ đông và sự kiện.

Mỗi ô code chạy độc lập và gọi dữ liệu thật từ API. Sau khi chạy, ô sẽ kiểm tra kiểu `DataFrame` và hiện 5 dòng đầu.

## Thị trường

```python
from stockiq import stock
import pandas as pd

# Thông tin mã cổ phiếu
df = stock.get_stock_info(symbol="ACB")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Lịch sử OHLCV theo ngày
df = stock.get_history(symbol="ACB", interval="days", start="2024-01-02", end="2024-01-10")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Lịch sử giao dịch tự doanh
df = stock.get_proprietary_history(symbol="ACB", interval="days", start="2024-01-02", end="2024-01-10")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert {"symbol", "date"} <= set(df.columns)
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Tổng hợp giao dịch tự doanh
df = stock.get_proprietary_history_summary(symbol="ACB", interval="days")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert {"symbol", "type"} <= set(df.columns)
    assert set(df["type"]) <= {"total", "average"}
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Lịch sử cung cầu
df = stock.get_demand_history(symbol="ACB", interval="days", start="2024-01-02", end="2024-01-10")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert {"symbol", "date"} <= set(df.columns)
    assert pd.api.types.is_datetime64_any_dtype(df["date"])
df.head()
```

## Cổ đông

```python
from stockiq import stock
import pandas as pd

# Thông tin cổ đông
df = stock.get_shareholders_info(symbol="ACB")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert pd.api.types.is_datetime64_any_dtype(df["updateDate"])
    assert pd.api.types.is_datetime64_any_dtype(df["publicDate"])
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Cơ cấu cổ đông
df = stock.get_shareholders_structure(symbol="ACB")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Các công ty liên quan
df = stock.get_shareholders_relationship(symbol="ACB")
assert isinstance(df, pd.DataFrame), type(df)
df.head()
```

```python
from stockiq import stock
import pandas as pd

# Giao dịch nội bộ
df = stock.get_insider_trans(symbol="ACB")
assert isinstance(df, pd.DataFrame), type(df)
if not df.empty:
    assert all(pd.api.types.is_datetime64_any_dtype(df[col]) for col in ("startDate", "endDate", "publicDate"))
df.head()
```
